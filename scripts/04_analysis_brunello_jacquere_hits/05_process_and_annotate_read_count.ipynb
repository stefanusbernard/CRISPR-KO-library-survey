```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

#### Process and annotate read count data for Brunello

```python
# import read count data
read_count = pd.read_excel("../../data/sgrna_lfc_data/brunello_data/brunello_A375_Sanson_2018.xlsx", sheet_name="A375_mod_tracr raw reads", skiprows = 1)
read_count = read_count.rename(columns={'sgRNA Sequence':'spacer'})
read_count

# import library data
library_data = pd.read_csv("../../data/library_data/refined_library/broadgpp-brunello-library-contents_63K_refined.tsv", sep ="\t", header = None)
library_data.columns = ['sgRNA', 'spacer', 'gene']
display(library_data.shape)

# merge read count data with library data
read_count = pd.merge(read_count, library_data, how="left", on="spacer").set_index(['sgRNA', 'spacer', 'gene']).sort_values(by="sgRNA")
read_count

# remove any index row with NA value
read_count = read_count[~read_count.index.to_frame().isna().any(axis=1)]
read_count

# remove spacer index
read_count = read_count.droplevel('spacer')
read_count

# export as txt file with tab separated value
read_count.to_csv("../../data/read_count_data/brunello_read_count_63K_refined_annotated_A375_mod_tracr.txt", sep = "\t")
```

#### Export non-targeting control of Brunello library

```python
# import library data
library_data = pd.read_csv("../../data/library_data/refined_library/broadgpp-brunello-library-contents_63K_refined.tsv", sep ="\t", header = None)
library_data.columns = ['sgRNA', 'spacer', 'gene']
library_data

control = library_data[library_data["gene"] == "Non-Targeting Control"]
control = control[["sgRNA"]]
control.to_csv("../../data/read_count_data/neg_control_brunello_refined.txt", header=False, index=False)
```

#### Process and annotate read count data for Jacquere

```python
# import read count data
read_count = pd.read_csv("../../data/read_count_data/counts-EK20250627_Jacquere_allinone.txt", sep = "\t")
read_count = read_count.rename(columns={'Construct Barcode':'spacer'})
read_count = read_count[["spacer", "CP0082_pDNA", "A375_CP0082_repA", "A375_CP0082_repB"]]
display(read_count.shape)

# import library data
library_data = pd.read_csv("../../data/library_data/refined_library/Jacquere_PerGuideAnnotations_Quota4_63K_refined.tsv", sep ="\t", header = None)
library_data.columns = ['sgRNA', 'spacer', 'gene']
display(library_data.shape)

# merge read count data with library data
read_count = pd.merge(read_count, library_data, how="left", on="spacer").set_index(['sgRNA', 'spacer', 'gene']).sort_values(by="sgRNA")
read_count

# remove any index row with NA value
read_count = read_count[~read_count.index.to_frame().isna().any(axis=1)]
read_count

# remove spacer index
read_count = read_count.droplevel('spacer')
read_count

# export as txt file with tab separated value
read_count.to_csv("../../data/read_count_data/counts-EK20250627_A375_refined_annotated_Jacquere_allinone.txt", sep = "\t")
```

```python
# import library data
library_data = pd.read_csv("../../data/library_data/refined_library/Jacquere_PerGuideAnnotations_Quota4_63K_refined.tsv", sep ="\t", header = None)
library_data.columns = ['sgRNA', 'spacer', 'gene']
library_data

control = library_data[library_data["gene"] == "NO_SITE"]
control = control[["sgRNA"]]
control.to_csv("../../data/read_count_data/neg_control_jacquere_refined.txt", header=False, index=False)
```
