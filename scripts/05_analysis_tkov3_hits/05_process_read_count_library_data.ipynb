### Process sgRNA read count and library data
1. for TKOv3, integrate read count data with refined TKOv3 library
2. remove all rows without sgRNA sequences
3. export data in a new format

```python
import pandas as pd
```

```python
read_count = pd.read_csv("../../data/sgrna_lfc_data/tkov3_data/matrix-reads-by-gRNA-RPE1-drugZ.txt", sep="\t")
read_count['spacer'] = read_count['sgRNA'].str.extract(r'_(.*)$')

col = read_count.pop('spacer')
read_count.insert(1, 'spacer', col)
read_count
```

```python
# check that number of unique spacer sequence match with the number of rows
read_count["spacer"].nunique()
```

```python
library = pd.read_csv("../../data/library_data/refined_library/tkov3_guide_sequence_61K_refined.tsv", sep = "\t", header=None)
library = library.rename(columns={0: "sgRNA", 1: "spacer", 2: "gene"})
library["spacer"]
```

```python
# only select read count that has sgRNA sequences data from refined library
read_count = read_count[read_count['spacer'].isin(library["spacer"])]
read_count
```

```python
# remove the spacer column and export the file in a delimited format
# read_count.pop('spacer')
read_count.to_csv("../../data/sgrna_lfc_data/tkov3_data/refined_matrix-reads-by-gRNA-RPE1-drugZ.txt", sep = "\t", index = None)
```
