```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib_venn import venn2, venn3
```

```python
def venn_diagram(hits_1, hits_2, name_1, name_2, colors=("red","blue"), fontsize=12):
    hits_1_set = set(hits_1["jacquere_original_hits"])
    hits_2_set = set(hits_2["jacquere_refined_hits"])

    plt.figure(figsize=(5, 5))
    v = venn2([hits_1_set, hits_2_set], (str(name_1), str(name_2)), set_colors=colors)

    # set patch colors and alpha
    # for pid, color in [('10', colors[0]), ('01', colors[1])]:
    #     patch = v.get_patch_by_id(pid)
    #     if patch is not None:
    #         patch.set_color(color)
    #         patch.set_alpha(0.6)
    #         patch.set_edgecolor('k')

    # set label font sizes
    for text in v.set_labels:
        if text is not None:
            text.set_fontsize(fontsize)
    for text in v.subset_labels:
        if text is not None:
            text.set_fontsize(max(13, fontsize - 2))

    plt.title("Comparison of gene hits between " + str(name_1) + " and " + str(name_2), fontsize=fontsize + 2)
    plt.show()
```

```python
data = pd.read_csv("../../data/result_gene_hits_change/brunello_jacquere_A375_hits_change.tsv")
data
```

```python
brunello_original_hits = data[["brunello_original_hits"]]
brunello_original_hits = brunello_original_hits.dropna()

brunello_refined_hits = data[["brunello_refined_hits"]]
brunello_refined_hits = brunello_refined_hits.dropna()

jacquere_original_hits = data[["jacquere_original_hits"]]
jacquere_original_hits = jacquere_original_hits.dropna()

jacquere_refined_hits = data[["jacquere_refined_hits"]]
jacquere_refined_hits = jacquere_refined_hits.dropna()
```
